```python
import re

def analyze_risks(text, company_name):
    results = {
        "company": company_name,
        "operational_risks": [],
        "regulatory_risks": [],
        "technological_risks": [],
        "ai_references": [],
        "cryptocurrency_references": [],
        "ecosystem_expansion_references": []
    }

    # Define keywords/patterns for each category
    operational_keywords = [
        "operational risk", "business interruption", "system failure",
        "data breach", "cybersecurity incident", "fraud", "talent retention",
        "supply chain disruption", "reputation damage", "disaster recovery",
        "processing errors", "network outages", "employee misconduct"
    ]
    regulatory_keywords = [
        "regulatory risk", "compliance", "legal proceedings", "government investigation",
        "sanctions", "anti-money laundering", "AML", "know your customer", "KYC",
        "data privacy", "GDPR", "CCPA", "licensing requirements", "fines and penalties",
        "enforcement actions", "consumer protection laws"
    ]
    technological_keywords = [
        "technological risk", "software bugs", "hardware failure", "IT infrastructure",
        "legacy systems", "system scalability", "technology obsolescence",
        "new technologies", "integration challenges", "security vulnerabilities"
    ]
    ai_keywords = ["artificial intelligence", "AI", "machine learning", "ML", "generative AI"]
    cryptocurrency_keywords = ["cryptocurrency", "bitcoin", "blockchain", "digital assets"]
    ecosystem_expansion_keywords = ["ecosystem expansion", "new markets", "strategic initiatives", "acquisitions", "partnerships", "international expansion"]

    # Helper function to find and extract sentences/paragraphs containing keywords
    def find_references(text, keywords, category):
        found_items = []
        for keyword in keywords:
            # Use regex to find sentences containing the keyword (case-insensitive)
            # This pattern attempts to capture a reasonable chunk of text around the keyword
            pattern = re.compile(f'[^.!?]*{re.escape(keyword)}[^.!?]*[.!?]', re.IGNORECASE)
            for match in pattern.finditer(text):
                sentence = match.group(0).strip()
                if sentence not in found_items: # Avoid duplicates
                    found_items.append(sentence)
        return found_items

    # Extract references for each category
    results["operational_risks"] = find_references(text, operational_keywords, "operational_risks")
    results["regulatory_risks"] = find_references(text, regulatory_keywords, "regulatory_risks")
    results["technological_risks"] = find_references(text, technological_keywords, "technological_risks")
    results["ai_references"] = find_references(text, ai_keywords, "ai_references")
    results["cryptocurrency_references"] = find_references(text, cryptocurrency_keywords, "cryptocurrency_references")
    results["ecosystem_expansion_references"] = find_references(text, ecosystem_expansion_keywords, "ecosystem_expansion_references")

    return results

block_analysis = analyze_risks(block_text, "Block, Inc.")
paypal_analysis = analyze_risks(paypal_text, "Paypal Holdings, Inc.")

import json

print("\n--- Block, Inc. Risk Analysis ---")
print(json.dumps(block_analysis, indent=2))

print("\n--- Paypal Holdings, Inc. Risk Analysis ---")
print(json.dumps(paypal_analysis, indent=2))
```

```python
from pypdf import PdfReader

def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text() + "\n"
    return text

block_pdf_path = '/content/input/Block, Inc 2023.pdf'
paypal_pdf_path = '/content/input/Paypal 2023.pdf'

block_text = extract_text_from_pdf(block_pdf_path)
paypal_text = extract_text_from_pdf(paypal_pdf_path)

print("Extracted text from Block, Inc 2023.pdf (first 500 chars):\n", block_text[:500])
print("\nExtracted text from Paypal 2023.pdf (first 500 chars):\n", paypal_text[:500])
```
