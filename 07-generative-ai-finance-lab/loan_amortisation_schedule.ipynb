# Objective
Calculate and display a loan amortization schedule given an initial principal of $100,000, an annual interest rate of 12%, and a start date of January 1, 2024. The schedule should account for a series of payments made at various dates and amounts. Finally, summarize the audit findings including the final loan balance and whether the borrower was overcharged interest.

```python
from datetime import date

principal = 100000.00
annual_interest_rate = 0.12
start_date = date(2024, 1, 1)

# Updated payment log to match the specific audit scenario
payment_log = [
    {'date': date(2024, 1, 31), 'amount': 5000.00},
    {'date': date(2024, 2, 29), 'amount': 5000.00}, # Leap Day
    {'date': date(2024, 3, 31), 'amount': 5000.00}
]

print(f"Loan parameters initialized for Leap Year 2024.")
print(f"Initial Principal: ${principal:,.2f}")
```

```python
current_balance = principal
last_date = start_date
amortization_schedule = []

# Daily interest calculation as specified: Balance * (0.12 / 365) * Days_Elapsed
daily_interest_rate = annual_interest_rate / 365

for payment in payment_log:
    payment_date = payment['date']
    payment_amount = payment['amount']

    # Calculate actual days elapsed (crucial for Leap Year/Irregular dates)
    days_elapsed = (payment_date - last_date).days

    # Calculate interest accrued for the specific period
    interest_accrued = current_balance * daily_interest_rate * days_elapsed

    # Apply payment: Interest first, then Principal
    principal_reduction = payment_amount - interest_accrued
    current_balance -= principal_reduction

    amortization_schedule.append({
        'date': payment_date,
        'days_elapsed': days_elapsed,
        'interest_accrued': round(interest_accrued, 4), # Higher precision for audit
        'payment_amount': round(payment_amount, 2),
        'principal_reduction': round(principal_reduction, 4),
        'new_balance': round(current_balance, 4)
    })

    last_date = payment_date

print('Amortization audit calculated.')
print(f'Final Balance after 3 payments: ${current_balance:,.2f}')
```

```python
import pandas as pd

# Create DataFrame from the calculated schedule
df_amortization = pd.DataFrame(amortization_schedule)

# Format date column
df_amortization['date'] = pd.to_datetime(df_amortization['date']).dt.strftime('%Y-%m-%d')

# Format monetary columns as currency strings
monetary_cols = ['interest_accrued', 'payment_amount', 'principal_reduction', 'new_balance']
for col in monetary_cols:
    df_amortization[col] = df_amortization[col].apply(lambda x: f"${x:,.2f}")

# Rename columns for better readability
df_amortization.columns = ['Date', 'Days Elapsed', 'Interest Accrued', 'Payment Amount', 'Principal Reduction', 'Remaining Balance']

# Display the table
print("Loan Amortization Schedule (Audit Result):")
display(df_amortization)
```
